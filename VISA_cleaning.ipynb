```python
import pandas as pd

# file paths
number_path = r"C:\MSC\TB-2\Visual Analytics\coursework\datasets\2011_general_health_number.csv"
percent_path = r"C:\MSC\TB-2\Visual Analytics\coursework\datasets\2011_general_health_percent.csv"
output_path = r"C:\MSC\TB-2\Visual Analytics\coursework\datasets\2011_general_health_LAD_clean.csv"

# keep only LAD-level geographies
lad_prefixes = ("E06", "E07", "E08", "E09", "W06")

# read raw files
number_df = pd.read_csv(number_path, encoding="latin-1", header=None)
percent_df = pd.read_csv(percent_path, encoding="latin-1", header=None)


# combine split area name columns
number_df["area_name"] = number_df[[1,2,3]].bfill(axis=1).iloc[:,0]
percent_df["area_name"] = percent_df[[1,2,3]].bfill(axis=1).iloc[:,0]

# keep only needed columns
number_df = number_df[[0, "area_name", 4,5,6,7,8,9]].copy()
percent_df = percent_df[[0, "area_name", 4,5,6,7,8,9]].copy()

# rename columns
number_df.columns = [
    "area_code", "area_name", "total_persons", "n_very_good", "n_good","n_fair", "n_bad", "n_very_bad"
]

percent_df.columns = [
    "area_code", "area_name", "total_persons", "pct_very_good", "pct_good", "pct_fair", "pct_bad", "pct_very_bad"
]

# keep only LAD rows
number_df = number_df[number_df["area_code"].str[:3].isin(lad_prefixes)].copy()
percent_df = percent_df[percent_df["area_code"].str[:3].isin(lad_prefixes)].copy()

# convert counts columns to numeric
for col in ["total_persons", "n_very_good", "n_good", "n_fair", "n_bad", "n_very_bad"]:
    number_df[col] =pd.to_numeric(
        number_df[col].astype(str).str.replace(",","",regex=False), errors="coerce"
    )

# covert percentage columns to numeric
for col in ["pct_very_good", "pct_good", "pct_fair", "pct_bad", "pct_very_bad"]:
    percent_df[col] = pd.to_numeric(percent_df[col], errors="coerce")

# drop duplicate total_persons from percentage file before merging
percent_df = percent_df.drop(columns=["total_persons", "area_name"])

# merge numbers and percentage
combined_df = number_df.merge(percent_df, on="area_code", how="inner")
combined_df["n_bad_or_very_bad"] = combined_df["n_bad"] + combined_df["n_very_bad"]
combined_df["pct_bad_or_very_bad"] = combined_df["pct_bad"] + combined_df["pct_very_bad"]

# save cleaned file
combined_df.to_csv(output_path, index=False)

print("Saved as 2011_general_health_LAD_clean.csv")
print(combined_df.head())
print(combined_df.shape)
print(combined_df.isna().sum())
print("Total missing values:", combined_df.isna().sum().sum())
```

```python
file_path = r"C:\MSC\TB-2\Visual Analytics\coursework\datasets\2021_general_health_raw.csv"
output_path = r"C:\MSC\TB-2\Visual Analytics\coursework\datasets\2021_general_health_LAD_clean.csv"

health_df_2021 = pd.read_csv(file_path)

# Rename columns
health_df_2021 = health_df_2021.rename(columns={
    "date": "year",
    "geography": "area_name",
    "geography code": "area_code",
    "General health: Total: All usual residents": "total_health_2021",
    "General health: Very good health": "n_very_good_2021",
    "General health: Good health": "n_good_2021",
    "General health: Fair health": "n_fair_2021",
    "General health: Bad health": "n_bad_2021",
    "General health: Very bad health": "n_very_bad_2021"
})

# Keep needed columns
health_df_2021 = health_df_2021[
    [
        "year",
        "area_code",
        "area_name",
        "total_health_2021",
        "n_very_good_2021",
        "n_good_2021",
        "n_fair_2021",
        "n_bad_2021",
        "n_very_bad_2021"
    ]
].copy()

# convert numeric columns
numeric_cols = [
    "total_health_2021",
    "n_very_good_2021",
    "n_good_2021",
    "n_fair_2021",
    "n_bad_2021",
    "n_very_bad_2021"

]
for col in numeric_cols:
    health_df_2021[col] = pd.to_numeric(health_df_2021[col], errors="coerce")

#

# Create percentage columns
health_df_2021["pct_very_good_2021"] = (health_df_2021["n_very_good_2021"]/health_df_2021["total_health_2021"])*100
health_df_2021["pct_good_2021"] = (health_df_2021["n_good_2021"]/health_df_2021["total_health_2021"])*100
health_df_2021["pct_fair_2021"] = (health_df_2021["n_fair_2021"]/health_df_2021["total_health_2021"])*100
health_df_2021["pct_bad_2021"] = (health_df_2021["n_bad_2021"]/health_df_2021["total_health_2021"])*100
health_df_2021["pct_very_bad_2021"] = (health_df_2021["n_very_bad_2021"]/health_df_2021["total_health_2021"])*100

# Main poor health measure
health_df_2021["n_bad_or_very_bad_2021"] = health_df_2021["n_bad_2021"] + health_df_2021["n_very_bad_2021"]
health_df_2021["pct_bad_or_very_bad_2021"] = (health_df_2021["n_bad_or_very_bad_2021"] / health_df_2021["total_health_2021"]) * 100

# Round percentage columns
pct_cols = [
    "pct_very_good_2021",
    "pct_good_2021",
    "pct_fair_2021",
    "pct_bad_2021",
    "pct_very_bad_2021",
    "pct_bad_or_very_bad_2021"
]
health_df_2021[pct_cols] = health_df_2021[pct_cols].round(2)

# final column order
health_df_2021 = health_df_2021[
    [
    "year",
    "area_code",
    "area_name",
    "total_health_2021",
    "n_very_good_2021",
    "n_good_2021",
    "n_fair_2021",
    "n_bad_2021",
    "n_very_bad_2021",
    "n_bad_or_very_bad_2021",
    "pct_very_good_2021",
    "pct_good_2021",
    "pct_fair_2021",
    "pct_bad_2021",
    "pct_very_bad_2021",
    "pct_bad_or_very_bad_2021",
    ]
]

# save clean dataset
health_df_2021.to_csv(output_path, index=False)

print(f"\nSaved as {output_path}")
print(health_df_2021.head())

print(health_df_2021.isna().sum())
print("Total missing values:", health_df_2021.isna().sum().sum())
```

```python
file_path = r"C:\MSC\TB-2\Visual Analytics\coursework\datasets\2021_economic_activity_raw.csv"
output_path = r"C:\MSC\TB-2\Visual Analytics\coursework\datasets\2021_economic_activity_LAD_clean.csv"

# read raw file
econ_df_2021 = pd.read_csv(file_path)

# rename columns
econ_df_2021 = econ_df_2021.rename(columns={
    "date": "year",
    "geography": "area_name",
    "geography code": "area_code",
    "Economic activity status: Total: All usual residents aged 16 years and over": "total_16plus_2021",
    "Economic activity status: Economically active (excluding full-time students)": "n_active_excl_students_2021",
    "Economic activity status: Economically active (excluding full-time students):In employment": "n_in_employment_excl_students_2021",
    "Economic activity status: Economically active (excluding full-time students): Unemployed": "n_unemployed_excl_students_2021",
    "Economic activity status: Economically active and a full-time student": "n_active_ft_student_2021",
    "Economic activity status: Economically active and a full-time student:In employment": "n_active_ft_student_in_employment_2021",
    "Economic activity status: Economically active and a full-time student: Unemployed": "n_active_ft_student_unemployed_2021",
    "Economic activity status: Economically inactive": "n_inactive_2021",
    "Economic activity status: Economically inactive: Retired": "n_retired_2021",
    "Economic activity status: Economically inactive: Student": "n_student_inactive_2021",
    "Economic activity status: Economically inactive: Looking after home or family": "n_home_family_2021",
    "Economic activity status: Economically inactive: Long-term sick or disabled": "n_long_term_sick_disabled_2021",
    "Economic activity status: Economically inactive: Other": "n_inactive_other_2021"
})

# keep needed columns
econ_df_2021 = econ_df_2021[
    [
        "year",
        "area_code",
        "area_name",
        "total_16plus_2021",
        "n_active_excl_students_2021",
        "n_in_employment_excl_students_2021",
        "n_unemployed_excl_students_2021",
        "n_active_ft_student_2021",
        "n_active_ft_student_in_employment_2021",
        "n_active_ft_student_unemployed_2021",
        "n_inactive_2021",
        "n_retired_2021",
        "n_student_inactive_2021",
        "n_home_family_2021",
        "n_long_term_sick_disabled_2021",
        "n_inactive_other_2021"
    ]
].copy()

# convert numeric columns
numeric_cols = [
    "total_16plus_2021",
    "n_active_excl_students_2021",
    "n_in_employment_excl_students_2021",
    "n_unemployed_excl_students_2021",
    "n_active_ft_student_2021",
    "n_active_ft_student_in_employment_2021",
    "n_active_ft_student_unemployed_2021",
    "n_inactive_2021",
    "n_retired_2021",
    "n_student_inactive_2021",
    "n_home_family_2021",
    "n_long_term_sick_disabled_2021",
    "n_inactive_other_2021"
]

for col in numeric_cols:
    econ_df_2021[col] = pd.to_numeric(econ_df_2021[col], errors="coerce")

# combined measures
econ_df_2021["n_active_total_2021"] = (
    econ_df_2021["n_active_excl_students_2021"] +
   econ_df_2021["n_active_ft_student_2021"]
)

econ_df_2021["n_in_employment_total_2021"] = (
    econ_df_2021["n_in_employment_excl_students_2021"] +
    econ_df_2021["n_active_ft_student_in_employment_2021"]
)

econ_df_2021["n_unemployed_total_2021"] = (
    econ_df_2021["n_unemployed_excl_students_2021"] +
   econ_df_2021["n_active_ft_student_unemployed_2021"]

)

# Percentage calculation

econ_df_2021["pct_active_total_2021"] = (econ_df_2021["n_active_total_2021"] / econ_df_2021["total_16plus_2021"]) * 100
econ_df_2021["pct_in_employment_total_2021"] = (econ_df_2021["n_in_employment_total_2021"] / econ_df_2021["total_16plus_2021"]) * 100
econ_df_2021["pct_unemployed_total_2021"] = (econ_df_2021["n_unemployed_total_2021"] / econ_df_2021["total_16plus_2021"]) * 100
econ_df_2021["pct_inactive_2021"] = (econ_df_2021["n_inactive_2021"] / econ_df_2021["total_16plus_2021"]) * 100
econ_df_2021["pct_retired_2021"] = (econ_df_2021["n_retired_2021"] / econ_df_2021["total_16plus_2021"]) * 100
econ_df_2021["pct_student_inactive_2021"] = (econ_df_2021["n_student_inactive_2021"] / econ_df_2021["total_16plus_2021"]) * 100
econ_df_2021["pct_home_family_2021"] = (econ_df_2021["n_home_family_2021"] / econ_df_2021["total_16plus_2021"]) * 100
econ_df_2021["pct_long_term_sick_disabled_2021"] = (econ_df_2021["n_long_term_sick_disabled_2021"] / econ_df_2021["total_16plus_2021"]) * 100
econ_df_2021["pct_inactive_other_2021"] = (econ_df_2021["n_inactive_other_2021"] / econ_df_2021["total_16plus_2021"]) * 100

pct_cols = [
    "pct_active_total_2021",
    "pct_in_employment_total_2021",
    "pct_unemployed_total_2021",
    "pct_inactive_2021",
    "pct_retired_2021",
    "pct_student_inactive_2021",
    "pct_home_family_2021",
    "pct_long_term_sick_disabled_2021",
    "pct_inactive_other_2021"
]
econ_df_2021[pct_cols] = econ_df_2021[pct_cols].round(2)

# final columns

econ_df_2021 = econ_df_2021[
    [
        "year",
        "area_code",
        "area_name",
        "total_16plus_2021",
        "n_active_excl_students_2021",
        "n_in_employment_excl_students_2021",
        "n_unemployed_excl_students_2021",
        "n_active_ft_student_2021",
        "n_active_ft_student_in_employment_2021",
        "n_active_ft_student_unemployed_2021",
        "n_active_total_2021",
        "n_in_employment_total_2021",
        "n_unemployed_total_2021",
        "n_inactive_2021",
        "n_retired_2021",
        "n_student_inactive_2021",
        "n_home_family_2021",
        "n_long_term_sick_disabled_2021",
        "n_inactive_other_2021",
        "pct_active_total_2021",
        "pct_in_employment_total_2021",
        "pct_unemployed_total_2021",
        "pct_inactive_2021",
        "pct_retired_2021",
        "pct_student_inactive_2021",
        "pct_home_family_2021",
        "pct_long_term_sick_disabled_2021",
        "pct_inactive_other_2021"
    ]
]

econ_df_2021.to_csv(output_path, index=False)

print(f"Saved as {output_path}")
print(econ_df_2021.head())

print(econ_df_2021.isna().sum())
print("Total missing values:", econ_df_2021.isna().sum().sum())
```
